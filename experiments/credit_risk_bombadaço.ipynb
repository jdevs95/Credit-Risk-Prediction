```python
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 200)
pd.set_option("display.width", 160)

df = pd.read_csv("C:\\Users\\Jorge Ferreira\\Desktop\\Credit Risk Prediction\\Loan_default.csv")

if "df" not in globals():
    raise NameError("The DataFrame `df` was not found in the notebook environment.")

data = df  # read-only alias; do not modify the original DataFrame
target_col = "Default"
id_col = "LoanID"

numeric_cols = [
    "Age", "Income", "LoanAmount", "CreditScore", "MonthsEmployed",
    "NumCreditLines", "InterestRate", "LoanTerm", "DTIRatio"
]

categorical_cols = [
    "Education", "EmploymentType", "MaritalStatus", "HasMortgage",
    "HasDependents", "LoanPurpose", "HasCoSigner"
]

all_feature_cols = numeric_cols + categorical_cols
profile_cols = [c for c in data.columns if c not in [id_col, target_col]]

default_numeric = pd.to_numeric(data[target_col], errors="coerce")

def section(title):
    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)

def note(text):
    print(f"Note: {text}")

def show(obj, text):
    display(obj)
    note(text)

def make_quantile_bins(series, q=4):
    s = series.dropna()
    if s.nunique() < 2:
        return pd.Series(index=series.index, dtype="object")
    try:
        bins = pd.qcut(s, q=min(q, s.nunique()), duplicates="drop")
        out = pd.Series(index=series.index, dtype="object")
        out.loc[s.index] = bins.astype(str)
        return out
    except Exception:
        try:
            bins = pd.cut(s, bins=min(q, max(2, s.nunique())), duplicates="drop")
            out = pd.Series(index=series.index, dtype="object")
            out.loc[s.index] = bins.astype(str)
            return out
        except Exception:
            return pd.Series(index=series.index, dtype="object")

def iqr_outlier_report(series):
    s = pd.to_numeric(series, errors="coerce").dropna()
    if s.empty:
        return {"q1": np.nan, "q3": np.nan, "iqr": np.nan, "lower": np.nan, "upper": np.nan, "outliers": 0}
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((s < lower) | (s > upper)).sum()
    return {"q1": q1, "q3": q3, "iqr": iqr, "lower": lower, "upper": upper, "outliers": int(outliers)}

# =============================================================================
# 1. Initial inspection
# =============================================================================
section("1. Initial inspection")

show(data.head(), "Displayed the first 5 rows to inspect the structure and sample records.")
show(data.sample(5, random_state=42), "Displayed 5 random rows for a broader spot-check of the data.")
print(f"Shape: {data.shape}")
note("Printed the dataset dimensions as (rows, columns).")
data.info()
note("Displayed the DataFrame information, including column types and non-null counts.")
show(data.dtypes.to_frame("dtype"), "Displayed the data types for each column.")
mem_bytes = data.memory_usage(deep=True).sum()
print(f"Total memory usage: {mem_bytes:,} bytes ({mem_bytes / 1024**2:.4f} MB)")
note("Printed the total memory footprint of the dataset.")

# =============================================================================
# 2. Descriptive statistics
# =============================================================================
section("2. Descriptive statistics")

num_desc = data[numeric_cols].describe().T
num_desc["median"] = data[numeric_cols].median(numeric_only=True)
num_desc["skewness"] = data[numeric_cols].skew(numeric_only=True)
num_desc["n_unique"] = data[numeric_cols].nunique(dropna=True)
show(num_desc, "Displayed numerical summary statistics, including skewness and unique-value counts.")

# Keep the requested object describe and also include the specified categorical columns safely
obj_desc = data.describe(include="object").T
show(obj_desc, 'Displayed the object-type summary using df.describe(include="object").T.')

cat_subset = data[categorical_cols].astype("object")
cat_desc = cat_subset.describe().T
cat_desc["category_proportions"] = cat_desc.apply(
    lambda row: data[row.name].astype("object").value_counts(normalize=True, dropna=False).to_dict(),
    axis=1
)
show(cat_desc, "Displayed categorical summaries for the specified categorical features.")

print("\nCategorical proportions by feature:")
for col in categorical_cols:
    print(f"\n{col}:")
    prop = data[col].astype("object").value_counts(normalize=True, dropna=False)
    display(prop.rename("proportion").to_frame())
    note(f"Displayed category proportions for {col}.")

# =============================================================================
# 3. Data quality
# =============================================================================
section("3. Data quality")

null_count = data.isna().sum().sort_values(ascending=False)
null_pct = (data.isna().mean() * 100).sort_values(ascending=False)
quality = pd.DataFrame({"null_count": null_count, "null_pct": null_pct})
show(quality, "Displayed null counts and null percentages for every column.")

exact_dupes = int(data.duplicated().sum())
print(f"Exact duplicate rows: {exact_dupes}")
note("Printed the number of exact duplicate rows.")

loanid_unique = data[id_col].is_unique
dup_loanid_count = int(data[id_col].duplicated().sum())
print(f"LoanID unique: {loanid_unique}")
print(f"Duplicated LoanID values (beyond the first occurrence): {dup_loanid_count}")
note("Reported whether LoanID is unique and how many duplicated LoanID values exist.")

dup_profiles = data.duplicated(subset=profile_cols).sum()
profile_groups = data.groupby(profile_cols, dropna=False)[target_col].nunique(dropna=False)
conflict_groups = int((profile_groups > 1).sum())
conflict_rows = int(
    data.groupby(profile_cols, dropna=False)[target_col]
    .transform(lambda s: s.nunique(dropna=False) > 1)
    .sum()
)
print(f"Duplicated borrower profile rows excluding {id_col} and {target_col}: {int(dup_profiles)}")
print(f"Duplicated borrower profile groups with conflicting Default values: {conflict_groups}")
print(f"Rows involved in conflicting profile groups: {conflict_rows}")
note("Reported duplicated borrower profiles and whether any duplicate profiles have conflicting targets.")

validation_checks = {
    "Default only 0/1": data[target_col].dropna().isin([0, 1]).all(),
    "Income non-negative": (pd.to_numeric(data["Income"], errors="coerce") >= 0).fillna(False).all(),
    "LoanAmount positive": (pd.to_numeric(data["LoanAmount"], errors="coerce") > 0).fillna(False).all(),
    "CreditScore plausible (300-850)": data["CreditScore"].dropna().between(300, 850).all(),
    "MonthsEmployed non-negative": (pd.to_numeric(data["MonthsEmployed"], errors="coerce") >= 0).fillna(False).all(),
    "NumCreditLines non-negative": (pd.to_numeric(data["NumCreditLines"], errors="coerce") >= 0).fillna(False).all(),
    "InterestRate non-negative": (pd.to_numeric(data["InterestRate"], errors="coerce") >= 0).fillna(False).all(),
    "LoanTerm positive": (pd.to_numeric(data["LoanTerm"], errors="coerce") > 0).fillna(False).all(),
    "DTIRatio within [0, 1]": data["DTIRatio"].dropna().between(0, 1).all(),
    "MonthsEmployed <= 600 months": (pd.to_numeric(data["MonthsEmployed"], errors="coerce") <= 600).fillna(False).all(),
}
validation_df = pd.DataFrame.from_dict(validation_checks, orient="index", columns=["passes"])
show(validation_df, "Displayed validation checks for required range and plausibility rules.")

print("\nInvalid / suspicious value counts:")
invalid_report = {
    "Default not in {0,1}": int((~data[target_col].dropna().isin([0, 1])).sum()),
    "Income < 0": int((pd.to_numeric(data["Income"], errors="coerce") < 0).sum()),
    "LoanAmount <= 0": int((pd.to_numeric(data["LoanAmount"], errors="coerce") <= 0).sum()),
    "CreditScore outside 300-850": int((~data["CreditScore"].dropna().between(300, 850)).sum()),
    "MonthsEmployed < 0": int((pd.to_numeric(data["MonthsEmployed"], errors="coerce") < 0).sum()),
    "NumCreditLines < 0": int((pd.to_numeric(data["NumCreditLines"], errors="coerce") < 0).sum()),
    "InterestRate < 0": int((pd.to_numeric(data["InterestRate"], errors="coerce") < 0).sum()),
    "LoanTerm <= 0": int((pd.to_numeric(data["LoanTerm"], errors="coerce") <= 0).sum()),
    "DTIRatio outside [0,1]": int((~data["DTIRatio"].dropna().between(0, 1)).sum()),
    "MonthsEmployed > 600": int((pd.to_numeric(data["MonthsEmployed"], errors="coerce") > 600).sum()),
}
invalid_df = pd.DataFrame.from_dict(invalid_report, orient="index", columns=["count"])
show(invalid_df, "Displayed counts of invalid or suspicious values without removing them.")

# =============================================================================
# 4. Target analysis
# =============================================================================
section("4. Target analysis")

default_counts = data[target_col].value_counts(dropna=False).sort_index()
default_props = data[target_col].value_counts(normalize=True, dropna=False).sort_index()
target_summary = pd.DataFrame({"count": default_counts, "proportion": default_props})
show(target_summary, "Displayed class counts and class proportions for the target variable.")

overall_default_rate = float(default_numeric.mean())
imbalance_ratio = np.nan
if 0 in default_counts.index and 1 in default_counts.index and default_counts.loc[1] > 0:
    imbalance_ratio = default_counts.loc[0] / default_counts.loc[1]

print(f"Overall default rate: {overall_default_rate:.6f}")
print(f"Class imbalance ratio (non-default / default): {imbalance_ratio:.4f}" if pd.notna(imbalance_ratio) else "Class imbalance ratio: not available.")
print(f"Baseline PR-AUC for a random ranking model: {overall_default_rate:.6f}")
print(f"Baseline Precision for a random classifier: {overall_default_rate:.6f}")
note("Printed the baseline performance expectations and class imbalance context.")

plt.figure(figsize=(6, 4))
sns.countplot(data=data, x=target_col)
plt.title("Default class count")
plt.tight_layout()
plt.show()
note("Displayed a count plot of the Default target.")

# =============================================================================
# 5. Numerical distributions
# =============================================================================
section("5. Numerical distributions")

num_corr = data[numeric_cols + [target_col]].corr(numeric_only=True)
show(num_corr, "Displayed the numerical correlation matrix, including the target.")

plt.figure(figsize=(11, 8))
sns.heatmap(num_corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Numerical correlation heatmap")
plt.tight_layout()
plt.show()
note("Displayed a heatmap of numerical correlations.")

corr_with_default = num_corr[target_col].drop(target_col).sort_values(key=lambda s: s.abs(), ascending=False)
show(corr_with_default.to_frame("corr_with_Default"), "Displayed correlations between each numerical feature and Default.")

num_outlier_summary = []
for col in numeric_cols:
    s = pd.to_numeric(data[col], errors="coerce")
    desc = s.describe()
    skew = s.skew()
    out = iqr_outlier_report(s)
    num_outlier_summary.append({
        "feature": col,
        "count": int(desc["count"]),
        "mean": desc["mean"],
        "std": desc["std"],
        "min": desc["min"],
        "25%": desc["25%"],
        "median": s.median(),
        "75%": desc["75%"],
        "max": desc["max"],
        "skewness": skew,
        "iqr_outliers": out["outliers"],
        "iqr_lower": out["lower"],
        "iqr_upper": out["upper"],
    })

num_outlier_df = pd.DataFrame(num_outlier_summary).set_index("feature")
show(num_outlier_df, "Displayed numerical distribution summaries, skewness, and IQR outlier counts.")

for col in numeric_cols:
    print(f"\n--- Numerical feature: {col} ---")
    s = pd.to_numeric(data[col], errors="coerce")
    by_target = data[[col, target_col]].copy()
    by_target[col] = pd.to_numeric(by_target[col], errors="coerce")
    group_stats = by_target.groupby(target_col)[col].agg(["count", "mean", "median", "std", "min", "max"])
    show(group_stats, f"Displayed distribution comparison for {col} by Default class.")

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.histplot(data=data, x=col, hue=target_col, kde=True, stat="density", common_norm=False, element="step", ax=axes[0])
    axes[0].set_title(f"{col} distribution by Default")
    sns.boxplot(data=data, x=target_col, y=col, ax=axes[1])
    axes[1].set_title(f"{col} box plot by Default")
    plt.tight_layout()
    plt.show()
    note(f"Displayed histogram and box plot for {col}.")

    bins = make_quantile_bins(s, q=4)
    band_df = pd.DataFrame({col: bins, target_col: default_numeric})
    band_df = band_df.dropna(subset=[col, target_col])
    if not band_df.empty:
        band_summary = band_df.groupby(col)[target_col].agg(["count", "sum", "mean"]).rename(columns={"sum": "defaults", "mean": "default_rate"})
        show(band_summary, f"Displayed default rate by quantile/meaningful band for {col}.")
    else:
        print(f"No valid banded summary could be created for {col}.")
        note(f"Skipped banded default-rate summary for {col} due to insufficient variation.")

    out = iqr_outlier_report(s)
    print(f"Skewness for {col}: {s.skew():.4f}")
    print(f"IQR outliers for {col}: {out['outliers']}")
    note(f"Reported skewness and outlier count for {col}.")

# Pair plot only if computationally reasonable
pair_cols = corr_with_default.abs().sort_values(ascending=False).head(min(4, len(numeric_cols))).index.tolist()
pair_plot_df = data[pair_cols + [target_col]].dropna()
if len(pair_plot_df) <= 5000 and len(pair_cols) >= 2:
    sample_n = min(500, len(pair_plot_df))
    sample_df = pair_plot_df.sample(sample_n, random_state=42) if len(pair_plot_df) > sample_n else pair_plot_df
    sns.pairplot(sample_df, vars=pair_cols, hue=target_col, corner=True, plot_kws={"alpha": 0.5, "s": 18})
    plt.suptitle("Pair plot of numerically strongest target-associated features", y=1.02)
    plt.show()
    note("Displayed a pair plot on a manageable sample using the strongest numerical associations with Default.")
else:
    note("Skipped pair plots because the dataset is not small enough for a practical pairwise visualization.")

# =============================================================================
# 6. Categorical distributions
# =============================================================================
section("6. Categorical distributions")

for col in categorical_cols:
    print(f"\n--- Categorical feature: {col} ---")
    series = data[col].astype("object")
    counts = series.value_counts(dropna=False)
    props = series.value_counts(normalize=True, dropna=False)
    cat_table = pd.DataFrame({"count": counts, "proportion": props})
    show(cat_table, f"Displayed category counts and proportions for {col}.")

    cat_default = (
        data[[col, target_col]]
        .copy()
        .assign(**{col: data[col].astype("object")})
        .groupby(col, dropna=False)[target_col]
        .agg(["count", "sum", "mean"])
        .rename(columns={"sum": "defaults", "mean": "default_rate"})
        .sort_values("default_rate", ascending=False)
    )
    show(cat_default, f"Displayed default rate per category for {col}.")

    plt.figure(figsize=(10, 4))
    sns.countplot(data=data, x=col, hue=target_col)
    plt.xticks(rotation=30, ha="right")
    plt.title(f"{col} count plot split by Default")
    plt.tight_layout()
    plt.show()
    note(f"Displayed a count plot split by Default for {col}.")

    plt.figure(figsize=(10, 4))
    plot_df = cat_default.reset_index()
    sns.barplot(data=plot_df, x=col, y="default_rate", color="steelblue")
    plt.xticks(rotation=30, ha="right")
    plt.title(f"{col} default-rate bar chart")
    plt.tight_layout()
    plt.show()
    note(f"Displayed a default-rate bar chart for {col}.")

    small_groups = cat_default[cat_default["count"] < 30]
    if not small_groups.empty:
        print(f"Small-sample categories in {col} (count < 30):")
        display(small_groups)
        note(f"Identified small sample-size categories for {col}.")
    else:
        note(f"No small sample-size categories were found for {col} using the count < 30 rule.")

    overall = overall_default_rate
    deviation = (cat_default["default_rate"] - overall).abs().sort_values(ascending=False)
    unusual = cat_default.loc[deviation.index]
    unusual = unusual[unusual["count"] >= 30].head(3)
    if not unusual.empty:
        print(f"Categories in {col} with the largest observed default-rate deviation from the overall rate (count >= 30):")
        display(unusual)
        note(f"Highlighted categories in {col} with unusually high or low default rates.")
    else:
        note(f"No category in {col} met the sample-size threshold for stable deviation analysis.")

# =============================================================================
# 7. Bivariate risk analysis
# =============================================================================
section("7. Bivariate risk analysis")

print("Numerical relationships are summarized in quantile/meaningful bands to avoid over-interpreting tiny groups.")
note("Bivariate risk analysis for numerical features will use banded groupings; categorical features use direct category grouping.")

for col in numeric_cols:
    s = pd.to_numeric(data[col], errors="coerce")
    bins = make_quantile_bins(s, q=4)
    tmp = pd.DataFrame({col: bins, target_col: default_numeric}).dropna(subset=[col, target_col])
    if tmp.empty:
        print(f"\n{col}: insufficient data for grouped analysis.")
        continue
    grp = tmp.groupby(col)[target_col].agg(group_size="count", defaults="sum", default_rate="mean")
    print(f"\n{col}:")
    display(grp)
    note(f"Reported group size, defaults, and default rate for {col}.")

for col in categorical_cols:
    tmp = data[[col, target_col]].copy()
    tmp[col] = tmp[col].astype("object")
    tmp[target_col] = default_numeric
    tmp = tmp.dropna(subset=[col, target_col])
    if tmp.empty:
        print(f"\n{col}: insufficient data for grouped analysis.")
        continue
    grp = tmp.groupby(col)[target_col].agg(group_size="count", defaults="sum", default_rate="mean").sort_values("default_rate", ascending=False)
    print(f"\n{col}:")
    display(grp)
    note(f"Reported group size, defaults, and default rate for {col}.")

# =============================================================================
# 8. EDA conclusions
# =============================================================================
section("8. EDA conclusions")

top_num_risks = corr_with_default.reindex(corr_with_default.abs().sort_values(ascending=False).index).head(5)

cat_risk_summary = []
for col in categorical_cols:
    tmp = data[[col, target_col]].copy()
    tmp[col] = tmp[col].astype("object")
    tmp[target_col] = default_numeric
    tmp = tmp.dropna(subset=[col, target_col])
    if tmp.empty:
        continue
    g = tmp.groupby(col)[target_col].agg(["count", "mean"])
    g = g[g["count"] >= 30]
    if not g.empty:
        max_dev = (g["mean"] - overall_default_rate).abs().max()
        cat_risk_summary.append((col, float(max_dev), g["mean"].idxmax(), int(g["count"].loc[g["mean"].idxmax()])))
cat_risk_summary = sorted(cat_risk_summary, key=lambda x: x[1], reverse=True)

missing_cols = quality[quality["null_count"] > 0].index.tolist()
invalid_nonzero = invalid_df[invalid_df["count"] > 0]

print(f"Dataset dimensions: {data.shape}")
print(f"Overall default rate: {overall_default_rate:.6f}")
print(f"Columns with missing values: {missing_cols if missing_cols else 'None observed'}")
print(f"Exact duplicate rows: {exact_dupes}")
print(f"Duplicated LoanID values: {dup_loanid_count}")
print(f"Duplicated borrower profile rows excluding LoanID and Default: {int(dup_profiles)}")
print(f"Conflicting duplicate profile groups: {conflict_groups}")
print("Invalid or suspicious values:")
display(invalid_nonzero if not invalid_nonzero.empty else pd.DataFrame({"count": []}))
note("Summarized the most important EDA findings.")

print("\nStrongest apparent numerical risk indicators (by absolute correlation with Default):")
display(top_num_risks.to_frame("corr_with_Default"))

if cat_risk_summary:
    print("\nStrongest apparent categorical risk indicators:")
    cat_risk_df = pd.DataFrame(
        cat_risk_summary,
        columns=["feature", "max_abs_deviation_from_overall", "category_with_max_rate", "category_sample_size"]
    ).set_index("feature")
    display(cat_risk_df)
else:
    note("No stable categorical risk indicator summary was available under the minimum sample-size rule.")

print("\nPotential outliers:")
outlier_focus = num_outlier_df[["iqr_outliers"]].sort_values("iqr_outliers", ascending=False)
display(outlier_focus)
note("Displayed numerical features with the largest IQR outlier counts.")

print("\nPotential multicollinearity:")
high_corr_pairs = []
corr_abs = num_corr[numeric_cols].abs()
for i, c1 in enumerate(numeric_cols):
    for c2 in numeric_cols[i+1:]:
        val = num_corr.loc[c1, c2]
        if abs(val) >= 0.80:
            high_corr_pairs.append((c1, c2, val))
high_corr_df = pd.DataFrame(high_corr_pairs, columns=["feature_1", "feature_2", "corr"]).sort_values("corr", key=lambda s: s.abs(), ascending=False)
display(high_corr_df if not high_corr_df.empty else pd.DataFrame({"message": ["No numerical feature pairs with |corr| >= 0.80 were found."]}))
note("Checked for strong numerical multicollinearity without removing any features.")

print("\nPossible fairness or proxy concerns:")
print("- Education, EmploymentType, MaritalStatus, HasMortgage, HasDependents, and LoanPurpose may act as sensitive or proxy-related variables.")
print("- Group-level default-rate differences should be interpreted carefully, especially for small categories.")
print("Possible leakage concerns:")
print("- InterestRate should be reviewed carefully because it may encode underwriting decisions or post-application pricing information.")
print("- Any feature derived from approval terms or post-decision processes could leak target-related information.")
print("Main limitations:")
print("- Observational data only; associations do not imply causation.")
print("- Small category sizes can make default rates unstable.")
print("- Binned numerical comparisons reduce granularity and may hide nonlinear effects.")
print("- Correlation analysis does not capture interaction effects or temporal causality.")
note("Printed concise dataset-level EDA conclusions without fitting any model.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import KBinsDiscretizer

# =========================
# 1) Original feature groups
# =========================
original_numerical_features = [
    "Age", "Income", "LoanAmount", "CreditScore", "MonthsEmployed",
    "NumCreditLines", "InterestRate", "LoanTerm", "DTIRatio"
]
original_categorical_features = [
    "Education", "EmploymentType", "MaritalStatus", "HasMortgage",
    "HasDependents", "LoanPurpose", "HasCoSigner"
]

target_name = "Default"
id_name = "LoanID"

# Candidate predictors (all original features excluding target and identifier)
X = data[original_numerical_features + original_categorical_features].copy()
y = data[target_name].copy()
loan_id = data[id_name].copy()

# =========================
# 2) Train-test split (leakage-safe)
# =========================
X_train, X_test, y_train, y_test, loanid_train, loanid_test = train_test_split(
    X, y, loan_id,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("\ny_train distribution:")
display(pd.DataFrame({"count": y_train.value_counts().sort_index(), "proportion": y_train.value_counts(normalize=True).sort_index()}))
print("y_test distribution:")
display(pd.DataFrame({"count": y_test.value_counts().sort_index(), "proportion": y_test.value_counts(normalize=True).sort_index()}))

# =========================
# 3–5,7) Deterministic FE (row-wise, no target/global stats)
# =========================
class DeterministicLoanFeatureEngineer(BaseEstimator, TransformerMixin):
    """
    Deterministic, row-wise feature engineering.
    - Does not use target.
    - Does not learn dataset-level parameters.
    - Payment formula is an approximation of monthly amortizing repayment.
    - PostLoanDTIProxy may double-count proposed loan if DTIRatio already includes it.
    - Several thresholds/bands are prototype assumptions.
    """
    def fit(self, X, y=None):
        return self

    @staticmethod
    def _safe_div(n, d):
        d = pd.to_numeric(d, errors="coerce")
        n = pd.to_numeric(n, errors="coerce")
        return n / d.replace(0, np.nan)

    @staticmethod
    def _yes_flag(s):
        return (
            s.astype("string")
             .str.strip()
             .str.lower()
             .isin(["yes", "y", "true", "1"])
             .astype("int8")
        )

    def transform(self, X):
        Xf = X.copy()

        # Ensure numeric columns are numeric for arithmetic
        for c in original_numerical_features:
            Xf[c] = pd.to_numeric(Xf[c], errors="coerce")

        # Income and exposure
        Xf["LogIncome"] = np.log1p(Xf["Income"])
        Xf["LogLoanAmount"] = np.log1p(Xf["LoanAmount"])
        Xf["MonthlyIncome"] = self._safe_div(Xf["Income"], 12)
        Xf["EmploymentYears"] = self._safe_div(Xf["MonthsEmployed"], 12)
        Xf["LoanTermYears"] = self._safe_div(Xf["LoanTerm"], 12)
        Xf["LoanToIncome"] = self._safe_div(Xf["LoanAmount"], Xf["Income"])
        Xf["LoanToTermIncome"] = self._safe_div(Xf["LoanAmount"], Xf["MonthlyIncome"] * Xf["LoanTerm"])

        # Estimated payment (approx monthly amortizing loan)
        P = Xf["LoanAmount"]
        r = self._safe_div(Xf["InterestRate"], 1200)  # InterestRate / 100 / 12
        n = Xf["LoanTerm"]

        pow_term = (1 + r) ** n
        denom = pow_term - 1

        amort_payment = np.where(
            (r > 0) & (denom != 0) & (n > 0),
            P * r * pow_term / denom,
            np.nan
        )
        zero_rate_payment = np.where(n > 0, self._safe_div(P, n), np.nan)

        Xf["MonthlyInterestRate"] = r
        Xf["EstimatedMonthlyPayment"] = np.where(r == 0, zero_rate_payment, amort_payment)
        Xf["TotalRepayment"] = Xf["EstimatedMonthlyPayment"] * n
        Xf["TotalInterest"] = Xf["TotalRepayment"] - P
        Xf["InterestBurden"] = self._safe_div(Xf["TotalInterest"], P)
        Xf["TotalInterestToIncome"] = self._safe_div(Xf["TotalInterest"], Xf["Income"])

        # Affordability and debt burden
        Xf["PaymentToIncome"] = self._safe_div(Xf["EstimatedMonthlyPayment"], Xf["MonthlyIncome"])
        Xf["AnnualPaymentToIncome"] = self._safe_div(Xf["EstimatedMonthlyPayment"] * 12, Xf["Income"])
        Xf["ExistingDebtPaymentProxy"] = Xf["DTIRatio"] * Xf["MonthlyIncome"]
        Xf["PostLoanDTIProxy"] = Xf["DTIRatio"] + Xf["PaymentToIncome"]
        Xf["ResidualIncomeProxy"] = Xf["MonthlyIncome"] - Xf["ExistingDebtPaymentProxy"] - Xf["EstimatedMonthlyPayment"]
        Xf["ResidualIncomeRatio"] = self._safe_div(Xf["ResidualIncomeProxy"], Xf["MonthlyIncome"])
        Xf["CreditBurden"] = Xf["LoanToIncome"] * Xf["DTIRatio"]

        # Employment stability
        Xf["AdultMonths"] = np.maximum((Xf["Age"] - 18) * 12, 1)
        Xf["EmploymentAdultLifeRatio"] = self._safe_div(Xf["MonthsEmployed"], Xf["AdultMonths"])
        Xf["EmploymentToLoanTermRatio"] = self._safe_div(Xf["MonthsEmployed"], Xf["LoanTerm"])
        Xf["LoanEndAge"] = Xf["Age"] + Xf["LoanTermYears"]
        Xf["LoanExtendsPastRetirement"] = (Xf["LoanEndAge"] > 65).astype("int8")
        Xf["EmploymentHistoryInconsistent"] = (Xf["MonthsEmployed"] > Xf["AdultMonths"]).astype("int8")

        # Credit-line exposure
        cl = Xf["NumCreditLines"].clip(lower=1)
        Xf["LoanPerCreditLine"] = self._safe_div(Xf["LoanAmount"], cl)
        Xf["IncomePerCreditLine"] = self._safe_div(Xf["Income"], cl)
        Xf["DTIPerCreditLine"] = self._safe_div(Xf["DTIRatio"], cl)
        Xf["ThinCreditFile"] = (Xf["NumCreditLines"] <= 1).astype("int8")

        # Binary and risk flags
        Xf["HasMortgageFlag"] = self._yes_flag(Xf["HasMortgage"])
        Xf["HasDependentsFlag"] = self._yes_flag(Xf["HasDependents"])
        Xf["HasCoSignerFlag"] = self._yes_flag(Xf["HasCoSigner"])
        Xf["ObligationCount"] = (Xf["HasMortgageFlag"] + Xf["HasDependentsFlag"]).astype("int8")

        # Prototype-threshold assumptions:
        Xf["HighBurdenNoCosigner"] = ((Xf["PaymentToIncome"] > 0.25) & (Xf["HasCoSignerFlag"] == 0)).astype("int8")
        Xf["LowScoreNoCosigner"] = ((Xf["CreditScore"] < 580) & (Xf["HasCoSignerFlag"] == 0)).astype("int8")
        Xf["HighDTIHighLoan"] = ((Xf["DTIRatio"] > 0.40) & (Xf["LoanToIncome"] > 1)).astype("int8")
        Xf["UnstableEmploymentLongTerm"] = ((Xf["MonthsEmployed"] < 12) & (Xf["LoanTerm"] >= 60)).astype("int8")

        # Fixed-domain bucketing (categorical)
        Xf["AgeBand"] = pd.cut(
            Xf["Age"],
            bins=[18, 25, 35, 45, 55, 65, np.inf],
            labels=["18–24", "25–34", "35–44", "45–54", "55–64", "65+"],
            right=False
        )

        Xf["CreditScoreBand"] = pd.cut(
            Xf["CreditScore"],
            bins=[-np.inf, 580, 670, 740, 800, np.inf],
            labels=["Poor", "Fair", "Good", "Very good", "Excellent"],
            right=False
        )

        Xf["EmploymentTenureBand"] = pd.cut(
            Xf["MonthsEmployed"],
            bins=[-np.inf, 6, 12, 24, 60, 120, np.inf],
            labels=["Less than 6 months", "6–11 months", "12–23 months", "24–59 months", "60–119 months", "120+ months"],
            right=False
        )

        # Prototype assumptions for bands not explicitly specified
        Xf["InterestRateBand"] = pd.cut(
            Xf["InterestRate"],
            bins=[-np.inf, 5, 10, 15, 20, np.inf],
            labels=["<5%", "5–9.99%", "10–14.99%", "15–19.99%", "20%+"],
            right=False
        )
        Xf["LoanTermBand"] = pd.cut(
            Xf["LoanTerm"],
            bins=[-np.inf, 24, 36, 48, 60, np.inf],
            labels=["<=24", "25–36", "37–48", "49–60", "60+"],
            right=False
        )
        Xf["DTIBand"] = pd.cut(
            Xf["DTIRatio"],
            bins=[-np.inf, 0.20, 0.35, 0.43, 0.60, np.inf],
            labels=["Below 0.20", "0.20–0.34", "0.35–0.42", "0.43–0.59", "0.60+"],
            right=False
        )
        Xf["LoanToIncomeBand"] = pd.cut(
            Xf["LoanToIncome"],
            bins=[-np.inf, 0.5, 1.0, 1.5, 2.0, np.inf],
            labels=["Below 0.50", "0.50–0.99", "1.00–1.49", "1.50–1.99", "2.00+"],
            right=False
        )
        Xf["PaymentToIncomeBand"] = pd.cut(
            Xf["PaymentToIncome"],
            bins=[-np.inf, 0.10, 0.20, 0.30, 0.40, np.inf],
            labels=["Below 0.10", "0.10–0.19", "0.20–0.29", "0.30–0.39", "0.40+"],
            right=False
        )

        # Categorical interactions (only requested ones)
        s = lambda c: Xf[c].astype("string").fillna("Missing")
        Xf["EmploymentEducation"] = s("EmploymentType") + " | " + s("Education")
        Xf["PurposeEmployment"] = s("LoanPurpose") + " | " + s("EmploymentType")
        Xf["PurposeMortgage"] = s("LoanPurpose") + " | " + s("HasMortgage")
        Xf["DependentsCosigner"] = s("HasDependents") + " | " + s("HasCoSigner")
        Xf["CreditBandCosigner"] = s("CreditScoreBand") + " | " + s("HasCoSigner")
        Xf["LoanBurdenEmployment"] = s("LoanToIncomeBand") + " | " + s("EmploymentType")

        # Replace infinities with missing so downstream pipeline can impute
        Xf = Xf.replace([np.inf, -np.inf], np.nan)
        return Xf


# =========================
# 6) Data-dependent bucketing (fit on train only)
# =========================
class QuantileBandTransformer(BaseEstimator, TransformerMixin):
    """
    Fold-safe quantile binner using KBinsDiscretizer.
    Fit on training data only; apply to train/test without refitting.
    """
    def __init__(self, cols=("Income", "LoanAmount"), n_bins=5):
        self.cols = list(cols)
        self.n_bins = n_bins
        self.kbin = KBinsDiscretizer(n_bins=n_bins, encode="ordinal", strategy="quantile")

    def fit(self, X, y=None):
        self.kbin.fit(X[self.cols])
        self.fit_rows_ = len(X)
        self.bin_edges_ = self.kbin.bin_edges_
        return self

    def transform(self, X):
        Xo = X.copy()
        binned = self.kbin.transform(Xo[self.cols])
        labels = [f"Q{i}" for i in range(1, self.n_bins + 1)]

        for i, c in enumerate(self.cols):
            idx = binned[:, i]
            out = pd.Series(pd.NA, index=Xo.index, dtype="object")
            valid = ~np.isnan(idx)
            out.loc[valid] = pd.Categorical.from_codes(idx[valid].astype(int), labels=labels, ordered=True).astype(str)
            Xo[f"{c}Band"] = out.astype("category")

        return Xo


# Apply deterministic FE separately to train and test
det_fe = DeterministicLoanFeatureEngineer()
# Patch _safe_div to handle both pandas objects and scalar denominators (e.g., 12)
def _safe_div_fixed(n, d):
    n_num = pd.to_numeric(n, errors="coerce")
    d_num = pd.to_numeric(d, errors="coerce")

    if np.isscalar(d_num):
        d_num = np.nan if d_num == 0 else d_num
    elif isinstance(d_num, (pd.Series, pd.DataFrame)):
        d_num = d_num.mask(d_num == 0)
    else:
        d_num = np.where(np.asarray(d_num) == 0, np.nan, d_num)

    return n_num / d_num

DeterministicLoanFeatureEngineer._safe_div = staticmethod(_safe_div_fixed)

X_train_det = det_fe.fit_transform(X_train)
X_test_det = det_fe.transform(X_test)

# Fit quantile boundaries on training only, then transform both
quant_fe = QuantileBandTransformer(cols=("Income", "LoanAmount"), n_bins=5)
def _quant_transform_fixed(self, X):
    Xo = X.copy()
    binned = self.kbin.transform(Xo[self.cols])
    categories = [f"Q{i}" for i in range(1, self.n_bins + 1)]

    for i, c in enumerate(self.cols):
        idx = binned[:, i]
        out = pd.Series(pd.NA, index=Xo.index, dtype="object")
        valid = ~np.isnan(idx)
        out.loc[valid] = pd.Categorical.from_codes(
            idx[valid].astype(int),
            categories=categories,   # pandas expects `categories`, not `labels`
            ordered=True
        ).astype(str)
        Xo[f"{c}Band"] = pd.Categorical(out, categories=categories, ordered=True)

    return Xo

QuantileBandTransformer.transform = _quant_transform_fixed
X_train_fe = quant_fe.fit_transform(X_train_det)
X_test_fe = quant_fe.transform(X_test_det)

# =========================
# 8) Feature sets for later comparison
# =========================
feature_set_A = ["Age", "Income", "LoanAmount", "CreditScore", "MonthsEmployed", "NumCreditLines", "InterestRate"]
feature_set_B = original_numerical_features + original_categorical_features
feature_set_C = X_train_fe.columns.tolist()  # complete engineered dataset

print("\nFeature set sizes:")
print("A:", len(feature_set_A))
print("B:", len(feature_set_B))
print("C:", len(feature_set_C))

# =========================
# 9) Validation
# =========================
print("\nPost-FE shapes:")
print("X_train_fe:", X_train_fe.shape)
print("X_test_fe :", X_test_fe.shape)

created_features = [c for c in X_train_fe.columns if c not in X_train.columns]
created_dtypes = X_train_fe[created_features].dtypes.astype(str).rename("dtype").to_frame()
print("\nCreated feature names and types:")
display(created_dtypes)

# Newly created null values
new_null_counts = X_train_fe[created_features].isna().sum().sort_values(ascending=False)
print("\nNewly created null values (train):")
display(new_null_counts[new_null_counts > 0] if (new_null_counts > 0).any() else pd.DataFrame({"message": ["No new nulls in created features."]}))

# Newly created infinite values
created_numeric = X_train_fe[created_features].select_dtypes(include=[np.number])
inf_count_train = np.isinf(created_numeric.to_numpy()).sum()
inf_count_test = np.isinf(X_test_fe[created_numeric.columns].to_numpy()).sum()
print(f"\nInfinite values in created numeric features -> train: {inf_count_train}, test: {inf_count_test}")

# Bucket distributions
bucket_cols = [
    "AgeBand", "CreditScoreBand", "EmploymentTenureBand", "InterestRateBand",
    "LoanTermBand", "DTIBand", "LoanToIncomeBand", "PaymentToIncomeBand",
    "IncomeBand", "LoanAmountBand"
]
print("\nBucket distributions (train, normalized):")
for c in bucket_cols:
    print(f"\n{c}:")
    display(X_train_fe[c].value_counts(normalize=True, dropna=False).rename("proportion").to_frame())

# Binary flags validation
binary_flags = [
    "HasMortgageFlag", "HasDependentsFlag", "HasCoSignerFlag", "ThinCreditFile",
    "LoanExtendsPastRetirement", "EmploymentHistoryInconsistent",
    "HighBurdenNoCosigner", "LowScoreNoCosigner", "HighDTIHighLoan",
    "UnstableEmploymentLongTerm"
]
print("\nBinary flag validation (unique values in train):")
for c in binary_flags:
    vals = sorted(pd.Series(X_train_fe[c].dropna().unique()).tolist())
    print(f"{c}: {vals}")

# Leakage/traceability checks
print("\nLeakage safety checks:")
print("Target not in modelling features:", target_name not in X_train_fe.columns)
print("LoanID not in modelling features:", id_name not in X_train_fe.columns)
print("X_test not used to learn quantile boundaries:", getattr(quant_fe, "fit_rows_", None) == len(X_train_det))
print("Quantile transformer learned from rows:", getattr(quant_fe, "fit_rows_", None))

# Feature catalogue
catalogue_rows = []

# Original features
for c in feature_set_B:
    catalogue_rows.append({
        "Feature name": c,
        "Feature family": "Original",
        "Formula or definition": "Original source feature",
        "Data type": str(X_train_fe[c].dtype),
        "Expected relationship with default risk": "Empirical; to be learned by model",
        "Main assumptions": "No transformation",
        "Potential leakage concerns": "Low (application-time field)"
    })

engineered_defs = {
    "LogIncome": ("Financial", "log1p(Income)", "Likely lower risk as income rises"),
    "LogLoanAmount": ("Financial", "log1p(LoanAmount)", "Potentially higher risk with larger exposure"),
    "MonthlyIncome": ("Financial", "Income / 12", "Higher monthly income may lower risk"),
    "EmploymentYears": ("Employment", "MonthsEmployed / 12", "Higher tenure may lower risk"),
    "LoanTermYears": ("Financial", "LoanTerm / 12", "Longer terms may increase uncertainty"),
    "LoanToIncome": ("Financial", "LoanAmount / Income", "Higher ratio may increase risk"),
    "LoanToTermIncome": ("Financial", "LoanAmount / (MonthlyIncome * LoanTerm)", "Higher ratio may increase risk"),
    "MonthlyInterestRate": ("Financial", "InterestRate / 100 / 12", "Higher cost may increase burden"),
    "EstimatedMonthlyPayment": ("Financial", "Amortizing payment approximation; if r=0 then LoanAmount/LoanTerm", "Higher payment burden may increase risk"),
    "TotalRepayment": ("Financial", "EstimatedMonthlyPayment * LoanTerm", "Higher total repay burden may increase risk"),
    "TotalInterest": ("Financial", "TotalRepayment - LoanAmount", "Higher interest paid may increase risk"),
    "InterestBurden": ("Financial", "TotalInterest / LoanAmount", "Higher burden may increase risk"),
    "TotalInterestToIncome": ("Financial", "TotalInterest / Income", "Higher ratio may increase risk"),
    "PaymentToIncome": ("Affordability", "EstimatedMonthlyPayment / MonthlyIncome", "Higher ratio may increase risk"),
    "AnnualPaymentToIncome": ("Affordability", "EstimatedMonthlyPayment * 12 / Income", "Higher ratio may increase risk"),
    "ExistingDebtPaymentProxy": ("Affordability", "DTIRatio * MonthlyIncome", "Higher existing burden may increase risk"),
    "PostLoanDTIProxy": ("Affordability", "DTIRatio + PaymentToIncome", "Higher total burden proxy may increase risk"),
    "ResidualIncomeProxy": ("Affordability", "MonthlyIncome - ExistingDebtPaymentProxy - EstimatedMonthlyPayment", "Lower residual may increase risk"),
    "ResidualIncomeRatio": ("Affordability", "ResidualIncomeProxy / MonthlyIncome", "Lower ratio may increase risk"),
    "CreditBurden": ("Affordability", "LoanToIncome * DTIRatio", "Higher burden may increase risk"),
    "AdultMonths": ("Employment", "max((Age-18)*12, 1)", "Support feature for consistency checks"),
    "EmploymentAdultLifeRatio": ("Employment", "MonthsEmployed / AdultMonths", "Lower ratio may imply instability"),
    "EmploymentToLoanTermRatio": ("Employment", "MonthsEmployed / LoanTerm", "Lower ratio may increase risk"),
    "LoanEndAge": ("Employment", "Age + LoanTermYears", "Extreme values may affect risk profile"),
    "LoanExtendsPastRetirement": ("Flag", "LoanEndAge > 65", "May increase repayment uncertainty"),
    "EmploymentHistoryInconsistent": ("Flag", "MonthsEmployed > AdultMonths", "Data-quality/risk signal"),
    "LoanPerCreditLine": ("Credit exposure", "LoanAmount / max(NumCreditLines,1)", "Higher per-line exposure may increase risk"),
    "IncomePerCreditLine": ("Credit exposure", "Income / max(NumCreditLines,1)", "Higher may reduce risk"),
    "DTIPerCreditLine": ("Credit exposure", "DTIRatio / max(NumCreditLines,1)", "Higher may increase risk"),
    "ThinCreditFile": ("Flag", "NumCreditLines <= 1", "May increase uncertainty/risk"),
    "HasMortgageFlag": ("Flag", "Binary map from HasMortgage", "Context dependent"),
    "HasDependentsFlag": ("Flag", "Binary map from HasDependents", "Potentially higher obligations"),
    "HasCoSignerFlag": ("Flag", "Binary map from HasCoSigner", "May reduce risk"),
    "ObligationCount": ("Flag", "HasMortgageFlag + HasDependentsFlag", "Higher obligations may increase risk"),
    "HighBurdenNoCosigner": ("Risk flag", "PaymentToIncome>0.25 and HasCoSignerFlag=0", "Likely higher risk"),
    "LowScoreNoCosigner": ("Risk flag", "CreditScore<580 and HasCoSignerFlag=0", "Likely higher risk"),
    "HighDTIHighLoan": ("Risk flag", "DTIRatio>0.40 and LoanToIncome>1", "Likely higher risk"),
    "UnstableEmploymentLongTerm": ("Risk flag", "MonthsEmployed<12 and LoanTerm>=60", "Likely higher risk"),
    "AgeBand": ("Bucket", "Fixed age bins", "Nonlinear age effects"),
    "CreditScoreBand": ("Bucket", "Fixed credit-score bins", "Captures credit quality tiers"),
    "EmploymentTenureBand": ("Bucket", "Fixed tenure bins", "Captures tenure nonlinearity"),
    "InterestRateBand": ("Bucket", "Prototype fixed bins", "Captures pricing nonlinearity"),
    "LoanTermBand": ("Bucket", "Prototype fixed bins", "Captures term nonlinearity"),
    "DTIBand": ("Bucket", "Fixed DTI bins", "Captures debt-burden tiers"),
    "LoanToIncomeBand": ("Bucket", "Prototype fixed bins", "Captures exposure tiers"),
    "PaymentToIncomeBand": ("Bucket", "Fixed affordability bins", "Captures burden tiers"),
    "IncomeBand": ("Quantile bucket", "KBinsDiscretizer quantile bins on TRAIN only", "Captures rank-based nonlinearity"),
    "LoanAmountBand": ("Quantile bucket", "KBinsDiscretizer quantile bins on TRAIN only", "Captures rank-based nonlinearity"),
    "EmploymentEducation": ("Categorical interaction", "EmploymentType + Education", "Joint profile effect"),
    "PurposeEmployment": ("Categorical interaction", "LoanPurpose + EmploymentType", "Joint profile effect"),
    "PurposeMortgage": ("Categorical interaction", "LoanPurpose + HasMortgage", "Joint profile effect"),
    "DependentsCosigner": ("Categorical interaction", "HasDependents + HasCoSigner", "Joint support/obligation effect"),
    "CreditBandCosigner": ("Categorical interaction", "CreditScoreBand + HasCoSigner", "Joint credit-support effect"),
    "LoanBurdenEmployment": ("Categorical interaction", "LoanToIncomeBand + EmploymentType", "Joint burden-stability effect"),
}

for feat, (fam, formula, expected) in engineered_defs.items():
    if feat in X_train_fe.columns:
        catalogue_rows.append({
            "Feature name": feat,
            "Feature family": fam,
            "Formula or definition": formula,
            "Data type": str(X_train_fe[feat].dtype),
            "Expected relationship with default risk": expected,
            "Main assumptions": "Prototype thresholds/bands where specified; deterministic row-wise construction",
            "Potential leakage concerns": "Low if built from application-time fields only"
        })

feature_catalogue = pd.DataFrame(catalogue_rows).sort_values(["Feature family", "Feature name"]).reset_index(drop=True)

print("\nFeature catalogue:")
display(feature_catalogue)
```

```python
from time import perf_counter
from collections import OrderedDict
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTENC
from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.model_selection import StratifiedKFold, GridSearchCV, cross_val_predict
from sklearn.inspection import permutation_importance
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

# SECTION 3 — MODEL DEVELOPMENT (leakage-safe SMOTENC pipelines)



from sklearn.metrics import (
    average_precision_score, roc_auc_score, precision_score, recall_score,
    f1_score, confusion_matrix, ConfusionMatrixDisplay, precision_recall_curve,
    roc_curve
)


# ---------------------------------------------------------------------
# 1. Fold-safe feature engineering wrapper
# ---------------------------------------------------------------------
class FoldSafeLoanFeatureEngineer(BaseEstimator, TransformerMixin):
    """Fits train-only quantile bands inside each CV fold."""

    def fit(self, X, y=None):
        self.det_ = DeterministicLoanFeatureEngineer()
        self.quant_ = QuantileBandTransformer(cols=("Income", "LoanAmount"), n_bins=5)

        X_det = self.det_.fit_transform(X, y)
        self.quant_.fit(X_det, y)
        return self

    def transform(self, X):
        return self.quant_.transform(self.det_.transform(X))


# Use the requested selected engineered features for the main comparison.
selected_feature_set = top10_features.copy()

feature_sets = OrderedDict({
    "Selected Top-10": selected_feature_set,
    "A: Numerical baseline": feature_set_A,
    "B: Original features": feature_set_B,
    "C: Full engineered": feature_set_C,
})

categorical_feature_pool = set(
    original_categorical_features
    + [
        "AgeBand", "CreditScoreBand", "EmploymentTenureBand", "InterestRateBand",
        "LoanTermBand", "DTIBand", "LoanToIncomeBand", "PaymentToIncomeBand",
        "IncomeBand", "LoanAmountBand", "EmploymentEducation",
        "PurposeEmployment", "PurposeMortgage", "DependentsCosigner",
        "CreditBandCosigner", "LoanBurdenEmployment"
    ]
)

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
k_values = [0.05, 0.10, 0.15, 0.20, 0.25, 0.30]
model_order = [
    "LogisticRegression", "LinearSVC", "DecisionTreeClassifier",
    "RandomForestClassifier", "LGBMClassifier", "XGBClassifier",
    "CatBoostClassifier", "AdaBoostClassifier"
]

model_specs = OrderedDict({
    "LogisticRegression": (
        LogisticRegression(max_iter=2000, solver="liblinear", random_state=42),
        {
            "model__C": [0.3, 1.0],
            "model__class_weight": [None, "balanced"],
        },
    ),
    "LinearSVC": (
        LinearSVC(max_iter=3000, random_state=42),
        {
            "model__C": [0.1, 1.0],
            "model__class_weight": [None, "balanced"],
        },
    ),
    "DecisionTreeClassifier": (
        DecisionTreeClassifier(random_state=42),
        {
            "model__max_depth": [4, 8],
            "model__min_samples_split": [20, 100],
            "model__min_samples_leaf": [10, 50],
            "model__ccp_alpha": [0.0, 0.001],
        },
    ),
    "RandomForestClassifier": (
        RandomForestClassifier(n_estimators=250, n_jobs=1, random_state=42),
        {
            "model__max_depth": [8, None],
            "model__min_samples_leaf": [5, 20],
            "model__max_features": ["sqrt", 0.7],
            "model__class_weight": [None, "balanced"],
        },
    ),
    "LGBMClassifier": (
        LGBMClassifier(
            objective="binary", n_jobs=1, random_state=42,
            verbosity=-1
        ),
        {
            "model__n_estimators": [250, 400],
            "model__learning_rate": [0.03, 0.08],
            "model__num_leaves": [15, 31],
            "model__min_child_samples": [30, 80],
            "model__subsample": [0.8],
            "model__colsample_bytree": [0.8],
            "model__reg_lambda": [0.0, 1.0],
        },
    ),
    "XGBClassifier": (
        XGBClassifier(
            objective="binary:logistic", eval_metric="logloss",
            n_jobs=1, random_state=42, tree_method="hist"
        ),
        {
            "model__n_estimators": [250, 400],
            "model__learning_rate": [0.03, 0.08],
            "model__max_depth": [3, 5],
            "model__min_child_weight": [5, 20],
            "model__subsample": [0.8],
            "model__colsample_bytree": [0.8],
            "model__reg_lambda": [1.0, 5.0],
        },
    ),
    "CatBoostClassifier": (
        CatBoostClassifier(
            loss_function="Logloss", verbose=0, thread_count=1,
            random_state=42, allow_writing_files=False
        ),
        {
            "model__iterations": [250, 400],
            "model__learning_rate": [0.03, 0.08],
            "model__depth": [4, 6],
            "model__l2_leaf_reg": [3.0, 8.0],
            "model__subsample": [0.8],
        },
    ),
    "AdaBoostClassifier": (
        AdaBoostClassifier(random_state=42),
        {
            "model__n_estimators": [150, 300],
            "model__learning_rate": [0.03, 0.1, 0.3],
        },
    ),
})


def make_pipeline(features, estimator):
    """Builds the required FE -> preprocessing -> SMOTENC -> OHE -> model pipeline."""
    categorical_features = [c for c in features if c in categorical_feature_pool]
    numeric_features = [c for c in features if c not in categorical_feature_pool]

    pre_smote = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="median")),
                    ("scaler", StandardScaler()),
                ]),
                numeric_features,
            ),
            (
                "cat",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    (
                        "ordinal",
                        OrdinalEncoder(
                            handle_unknown="use_encoded_value",
                            unknown_value=-1,
                        ),
                    ),
                ]),
                categorical_features,
            ),
        ],
        remainder="drop",
        verbose_feature_names_out=False,
    ).set_output(transform="pandas")

    n_numeric = len(numeric_features)
    categorical_indices = list(range(n_numeric, n_numeric + len(categorical_features)))

    post_smote = ColumnTransformer(
        transformers=[
            ("num", "passthrough", numeric_features),
            (
                "cat",
                OneHotEncoder(handle_unknown="ignore", sparse_output=False),
                categorical_features,
            ),
        ],
        verbose_feature_names_out=False,
    ).set_output(transform="pandas")

    return Pipeline([
        ("feature_engineering", FoldSafeLoanFeatureEngineer()),
        ("pre_smote", pre_smote),
        (
            "smote",
            SMOTENC(
                categorical_features=categorical_indices,
                sampling_strategy=0.5,
                random_state=42,
            ),
        ),
        ("post_smote", post_smote),
        ("model", estimator),
    ])


def get_scores(estimator, X):
    if hasattr(estimator, "predict_proba"):
        return estimator.predict_proba(X)[:, 1]
    return estimator.decision_function(X)


def choose_thresholds(y_true, scores, probability_model):
    default_threshold = 0.5 if probability_model else 0.0

    precision, recall, thresholds = precision_recall_curve(y_true, scores)
    f1_values = 2 * precision * recall / np.clip(precision + recall, 1e-12, None)

    if len(thresholds):
        f1_threshold = thresholds[np.nanargmax(f1_values[:-1])]
        valid_business = np.where(recall[:-1] >= 0.60)[0]
        business_threshold = (
            thresholds[valid_business[np.argmax(precision[:-1][valid_business])]]
            if len(valid_business) else f1_threshold
        )
    else:
        f1_threshold = default_threshold
        business_threshold = default_threshold

    return {
        "Default": float(default_threshold),
        "F1-optimized": float(f1_threshold),
        "Business (Recall ≥ 60%)": float(business_threshold),
    }


def top_k_metrics(y_true, scores):
    y_array = np.asarray(y_true)
    order = np.argsort(-np.asarray(scores))
    base_rate = y_array.mean()
    results = {}

    for k in k_values:
        n_flagged = max(1, int(np.ceil(len(y_array) * k)))
        selected = order[:n_flagged]
        captured = y_array[selected].sum()
        precision_k = captured / n_flagged
        recall_k = captured / max(y_array.sum(), 1)

        results[f"Precision@{int(k * 100)}%"] = precision_k
        results[f"Recall@{int(k * 100)}%"] = recall_k
        results[f"Lift@{int(k * 100)}%"] = precision_k / base_rate
        results[f"Flagged@{int(k * 100)}%"] = n_flagged
        results[f"DefaultsCaptured@{int(k * 100)}%"] = int(captured)

    return results


def evaluate_threshold(y_true, scores, threshold):
    pred = (np.asarray(scores) >= threshold).astype(int)
    return {
        "Precision": precision_score(y_true, pred, zero_division=0),
        "Recall": recall_score(y_true, pred, zero_division=0),
        "F1-score": f1_score(y_true, pred, zero_division=0),
        "Predicted-positive rate": pred.mean(),
        "Confusion matrix": confusion_matrix(y_true, pred),
    }


def run_experiment(model_name, features, feature_set_name):
    estimator, param_grid = model_specs[model_name]
    pipe = make_pipeline(features, clone(estimator))

    search = GridSearchCV(
        estimator=pipe,
        param_grid=param_grid,
        scoring="average_precision",
        cv=cv,
        refit=True,
        return_train_score=True,
        n_jobs=-1,
    )

    start = perf_counter()
    search.fit(X_train, y_train)
    grid_time = perf_counter() - start

    best_estimator = search.best_estimator_
    best_idx = search.best_index_
    cv_results = search.cv_results_

    score_method = "predict_proba" if hasattr(best_estimator, "predict_proba") else "decision_function"

    oof_scores = cross_val_predict(
        clone(best_estimator),
        X_train,
        y_train,
        cv=cv,
        method=score_method,
        n_jobs=-1,
    )
    if score_method == "predict_proba":
        oof_scores = oof_scores[:, 1]

    thresholds = choose_thresholds(
        y_train,
        oof_scores,
        probability_model=(score_method == "predict_proba"),
    )

    test_scores = get_scores(best_estimator, X_test)
    roc_auc = roc_auc_score(y_test, test_scores)
    pr_auc = average_precision_score(y_test, test_scores)
    ranking = top_k_metrics(y_test, test_scores)

    feature_names = best_estimator.named_steps["post_smote"].get_feature_names_out()
    n_transformed_features = len(feature_names)

    metric_rows = []
    for threshold_type, threshold in thresholds.items():
        metrics = evaluate_threshold(y_test, test_scores, threshold)
        metric_rows.append({
            "Model": model_name,
            "Feature set": feature_set_name,
            "Threshold type": threshold_type,
            "Selected threshold": threshold,
            "Precision": metrics["Precision"],
            "Recall": metrics["Recall"],
            "F1-score": metrics["F1-score"],
            "ROC-AUC": roc_auc,
            "PR-AUC": pr_auc,
            "Predicted-positive rate": metrics["Predicted-positive rate"],
            "Best CV PR-AUC": search.best_score_,
            "CV PR-AUC standard deviation": cv_results["std_test_score"][best_idx],
            "Training time": grid_time,
            "Number of transformed features": n_transformed_features,
            **ranking,
        })

    params_row = {
        "Model": model_name,
        "Feature set": feature_set_name,
        "Best hyperparameters": search.best_params_,
        "Best CV PR-AUC": search.best_score_,
        "Mean training CV PR-AUC": cv_results["mean_train_score"][best_idx],
        "Mean validation CV PR-AUC": cv_results["mean_test_score"][best_idx],
        "CV standard deviation": cv_results["std_test_score"][best_idx],
        "Mean fit time": cv_results["mean_fit_time"][best_idx],
        "Total Grid Search time": grid_time,
        "Number of transformed features": n_transformed_features,
    }

    print(f"\n{'=' * 90}")
    print(f"{model_name} | {feature_set_name}")
    print(f"Best CV PR-AUC: {search.best_score_:.4f}")
    print(f"Test PR-AUC: {pr_auc:.4f} | Test ROC-AUC: {roc_auc:.4f}")
    print(f"Best parameters: {search.best_params_}")
    print(f"Grid-search time: {grid_time:.1f} seconds")

    for row in metric_rows:
        print(
            f"{row['Threshold type']}: threshold={row['Selected threshold']:.4f}, "
            f"Precision={row['Precision']:.4f}, Recall={row['Recall']:.4f}, "
            f"F1={row['F1-score']:.4f}"
        )

    return {
        "estimator": best_estimator,
        "features": features,
        "feature_names": feature_names,
        "test_scores": test_scores,
        "thresholds": thresholds,
        "metric_rows": metric_rows,
        "params_row": params_row,
    }


# ---------------------------------------------------------------------
# 2. Main model comparison using the requested Top-10 engineered features
# ---------------------------------------------------------------------
master_rows = []
parameter_rows = []
experiment_store = {}

for model_name in model_order:
    result = run_experiment(
        model_name=model_name,
        features=selected_feature_set,
        feature_set_name="Selected Top-10",
    )
    experiment_store[(model_name, "Selected Top-10")] = result
    master_rows.extend(result["metric_rows"])
    parameter_rows.append(result["params_row"])

master_metrics = pd.DataFrame(master_rows)
best_parameters = pd.DataFrame(parameter_rows)

display(
    master_metrics.sort_values(
        ["PR-AUC", "F1-score"], ascending=False
    ).reset_index(drop=True)
)
display(best_parameters.sort_values("Best CV PR-AUC", ascending=False).reset_index(drop=True))

# ---------------------------------------------------------------------
# 3. Feature-set ablation for the top three primary models
# ---------------------------------------------------------------------
top_three_models = (
    master_metrics[master_metrics["Threshold type"] == "F1-optimized"]
    .sort_values("PR-AUC", ascending=False)["Model"]
    .head(3)
    .tolist()
)

for model_name in top_three_models:
    for feature_set_name, features in feature_sets.items():
        if feature_set_name == "Selected Top-10":
            continue

        result = run_experiment(model_name, features, feature_set_name)
        experiment_store[(model_name, feature_set_name)] = result
        master_rows.extend(result["metric_rows"])
        parameter_rows.append(result["params_row"])

master_metrics = pd.DataFrame(master_rows)
best_parameters = pd.DataFrame(parameter_rows)

ablation_metrics = (
    master_metrics[master_metrics["Threshold type"] == "F1-optimized"]
    .query("Model in @top_three_models")
    [[
        "Model", "Feature set", "PR-AUC", "ROC-AUC", "F1-score",
        "Precision@10%", "Recall@10%", "Lift@10%", "Training time"
    ]]
    .sort_values(["Model", "PR-AUC"], ascending=[True, False])
)

print("\nFeature-set ablation:")
display(ablation_metrics)

# Incremental comparisons relative to numerical baseline.
ablation_pivot = ablation_metrics.pivot(index="Model", columns="Feature set")
baseline_label = "A: Numerical baseline"

if baseline_label in ablation_pivot.columns.get_level_values(1):
    incremental_performance = pd.DataFrame(index=ablation_pivot.index)
    for metric in ["PR-AUC", "ROC-AUC", "F1-score", "Precision@10%", "Recall@10%", "Lift@10%"]:
        baseline = ablation_pivot[(metric, baseline_label)]
        for compare_label in ["B: Original features", "C: Full engineered"]:
            if (metric, compare_label) in ablation_pivot.columns:
                incremental_performance[f"{metric}: {compare_label} minus A"] = (
                    ablation_pivot[(metric, compare_label)] - baseline
                )

    print("\nIncremental feature-set performance relative to Feature Set A:")
    display(incremental_performance)

# ---------------------------------------------------------------------
# 4. Interest-rate leakage experiment for top three models
# ---------------------------------------------------------------------
interest_rate_derived = {
    "InterestRate",
    "InterestRateBand",
    "MonthlyInterestRate",
    "EstimatedMonthlyPayment",
    "TotalRepayment",
    "TotalInterest",
    "InterestBurden",
    "TotalInterestToIncome",
    "PaymentToIncome",
    "AnnualPaymentToIncome",
    "PostLoanDTIProxy",
    "ResidualIncomeProxy",
    "ResidualIncomeRatio",
    "HighBurdenNoCosigner",
    "PaymentToIncomeBand",
}

booked_features = feature_set_C.copy()
pre_pricing_features = [f for f in feature_set_C if f not in interest_rate_derived]

for model_name in top_three_models:
    for variant_name, variant_features in {
        "Booked-loan (InterestRate included)": booked_features,
        "Pre-pricing (InterestRate excluded)": pre_pricing_features,
    }.items():
        result = run_experiment(model_name, variant_features, variant_name)
        experiment_store[(model_name, variant_name)] = result
        master_rows.extend(result["metric_rows"])
        parameter_rows.append(result["params_row"])

master_metrics = pd.DataFrame(master_rows)
best_parameters = pd.DataFrame(parameter_rows)

interest_experiment = (
    master_metrics[
        master_metrics["Feature set"].isin([
            "Booked-loan (InterestRate included)",
            "Pre-pricing (InterestRate excluded)",
        ])
        & (master_metrics["Threshold type"] == "F1-optimized")
    ]
    [[
        "Model", "Feature set", "PR-AUC", "ROC-AUC", "F1-score",
        "Precision@10%", "Recall@10%", "Lift@10%", "Training time"
    ]]
    .sort_values(["Model", "Feature set"])
)

print("\nInterest-rate leakage experiment:")
display(interest_experiment)

# ---------------------------------------------------------------------
# 5. Charts: primary selected Top-10 model comparison
# ---------------------------------------------------------------------
primary_metrics = master_metrics[
    (master_metrics["Feature set"] == "Selected Top-10")
    & (master_metrics["Threshold type"] == "F1-optimized")
].copy()

primary_metrics["Model"] = pd.Categorical(
    primary_metrics["Model"], categories=model_order, ordered=True
)
primary_metrics = primary_metrics.sort_values("Model")

palette = dict(zip(model_order, sns.color_palette("tab10", n_colors=len(model_order))))

plot_columns = ["Precision", "Recall", "F1-score", "ROC-AUC", "PR-AUC"]
plot_long = primary_metrics.melt(
    id_vars="Model", value_vars=plot_columns,
    var_name="Metric", value_name="Score"
)

plt.figure(figsize=(15, 6))
sns.barplot(data=plot_long, x="Model", y="Score", hue="Metric")
plt.ylim(0, 1)
plt.xticks(rotation=35, ha="right")
plt.title("Selected Top-10: Classification Metric Comparison")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
for model_name in primary_metrics["Model"]:
    result = experiment_store[(model_name, "Selected Top-10")]
    fpr, tpr, _ = roc_curve(y_test, result["test_scores"])
    plt.plot(
        fpr, tpr,
        label=f"{model_name} (AUC={roc_auc_score(y_test, result['test_scores']):.3f})",
        color=palette[model_name],
    )
plt.plot([0, 1], [0, 1], "k--", lw=1)
plt.xlabel("False-positive rate")
plt.ylabel("True-positive rate")
plt.title("ROC Curve Comparison")
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
for model_name in primary_metrics["Model"]:
    result = experiment_store[(model_name, "Selected Top-10")]
    precision, recall, _ = precision_recall_curve(y_test, result["test_scores"])
    plt.plot(
        recall, precision,
        label=f"{model_name} (AP={average_precision_score(y_test, result['test_scores']):.3f})",
        color=palette[model_name],
    )
plt.axhline(y_test.mean(), color="k", linestyle="--", lw=1, label="Base default rate")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve Comparison")
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()

for metric_prefix, title in [
    ("Precision", "Precision@k"),
    ("Recall", "Recall@k"),
    ("Lift", "Lift@k"),
]:
    plt.figure(figsize=(9, 5))
    for _, row in primary_metrics.iterrows():
        model_name = row["Model"]
        values = [row[f"{metric_prefix}@{int(k * 100)}%"] for k in k_values]
        plt.plot(
            [int(k * 100) for k in k_values],
            values,
            marker="o",
            label=model_name,
            color=palette[model_name],
        )
    plt.xlabel("Top-k borrowers flagged (%)")
    plt.ylabel(metric_prefix)
    plt.title(f"{title} Curve Comparison")
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(10, 5))
sns.barplot(data=primary_metrics, x="Model", y="Training time", palette=palette)
plt.xticks(rotation=35, ha="right")
plt.title("Grid Search Training-Time Comparison")
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=primary_metrics,
    x="Best CV PR-AUC",
    y="PR-AUC",
    hue="Model",
    s=120,
    palette=palette,
)
plt.plot([0, 1], [0, 1], "k--", lw=1)
plt.title("Cross-Validation PR-AUC versus Test PR-AUC")
plt.tight_layout()
plt.show()

# Confusion matrices for every primary model at its F1-optimized threshold.
for _, row in primary_metrics.iterrows():
    model_name = row["Model"]
    result = experiment_store[(model_name, "Selected Top-10")]
    threshold = result["thresholds"]["F1-optimized"]
    predictions = (result["test_scores"] >= threshold).astype(int)

    fig, ax = plt.subplots(figsize=(4, 4))
    ConfusionMatrixDisplay(
        confusion_matrix(y_test, predictions),
        display_labels=["Non-default", "Default"],
    ).plot(ax=ax, colorbar=False)
    ax.set_title(f"{model_name}\nF1 threshold={threshold:.4f}")
    plt.tight_layout()
    plt.show()

# ---------------------------------------------------------------------
# 6. Feature interpretation and permutation importance
# ---------------------------------------------------------------------
permutation_rows = []
sample_size = min(5000, len(X_test))
test_sample_idx = (
    y_test.groupby(y_test, group_keys=False)
    .apply(lambda s: s.sample(
        n=max(1, int(sample_size * len(s) / len(y_test))),
        random_state=42,
    ))
    .index
)

X_test_perm = X_test.loc[test_sample_idx]
y_test_perm = y_test.loc[test_sample_idx]

for model_name in model_order:
    result = experiment_store[(model_name, "Selected Top-10")]
    fitted_pipeline = result["estimator"]
    fitted_model = fitted_pipeline.named_steps["model"]
    feature_names = result["feature_names"]

    # Transform test data through FE/preprocessing/final encoding only.
    X_fe = fitted_pipeline.named_steps["feature_engineering"].transform(X_test_perm)
    X_pre = fitted_pipeline.named_steps["pre_smote"].transform(X_fe)
    X_final = fitted_pipeline.named_steps["post_smote"].transform(X_pre)

    perm = permutation_importance(
        fitted_model,
        X_final,
        y_test_perm,
        scoring="average_precision",
        n_repeats=5,
        random_state=42,
        n_jobs=-1,
    )

    perm_df = pd.DataFrame({
        "Model": model_name,
        "Feature": feature_names,
        "Importance mean": perm.importances_mean,
        "Importance std": perm.importances_std,
    }).sort_values("Importance mean", ascending=False)

    permutation_rows.append(perm_df)

    top_perm = perm_df.head(20).sort_values("Importance mean")
    plt.figure(figsize=(9, 7))
    plt.barh(
        top_perm["Feature"],
        top_perm["Importance mean"],
        xerr=top_perm["Importance std"],
        color=palette[model_name],
        alpha=0.8,
    )
    plt.title(f"{model_name}: Permutation Importance (AP scoring)")
    plt.xlabel("Mean decrease in average precision")
    plt.tight_layout()
    plt.show()

    # Signed linear coefficients.
    if model_name in {"LogisticRegression", "LinearSVC"}:
        coef_df = pd.DataFrame({
            "Feature": feature_names,
            "Coefficient": fitted_model.coef_.ravel(),
        })
        coef_df["abs_coefficient"] = coef_df["Coefficient"].abs()
        coef_df = coef_df.nlargest(20, "abs_coefficient").sort_values("Coefficient")

        plt.figure(figsize=(9, 7))
        colors = np.where(coef_df["Coefficient"] >= 0, "firebrick", "steelblue")
        plt.barh(coef_df["Feature"], coef_df["Coefficient"], color=colors)
        plt.axvline(0, color="black", lw=1)
        plt.title(f"{model_name}: Signed Standardized Coefficients")
        plt.xlabel("Positive = increased predicted default risk")
        plt.tight_layout()
        plt.show()

    # Native tree/boosting importance. Definition depends on estimator:
    # RF/DecisionTree/AdaBoost = impurity-based; LGBM/XGBoost/CatBoost = gain-like native importance.
    elif hasattr(fitted_model, "feature_importances_"):
        native_df = pd.DataFrame({
            "Feature": feature_names,
            "Native importance": fitted_model.feature_importances_,
        }).sort_values("Native importance", ascending=False)

        top_native = native_df.head(20).sort_values("Native importance")
        plt.figure(figsize=(9, 7))
        plt.barh(top_native["Feature"], top_native["Native importance"], color=palette[model_name])
        plt.title(f"{model_name}: Native Feature Importance")
        plt.tight_layout()
        plt.show()

permutation_importance_results = pd.concat(permutation_rows, ignore_index=True)
display(
    permutation_importance_results
    .sort_values(["Model", "Importance mean"], ascending=[True, False])
    .groupby("Model", group_keys=False)
    .head(20)
)

# ---------------------------------------------------------------------
# 7. Final ranking and recommendations
# ---------------------------------------------------------------------
final_ranking = (
    master_metrics[master_metrics["Threshold type"] == "F1-optimized"]
    .sort_values(
        ["PR-AUC", "Recall@10%", "Precision@10%", "CV PR-AUC standard deviation"],
        ascending=[False, False, False, True],
    )
    .reset_index(drop=True)
)

print("\nFinal ranking (primarily by test PR-AUC):")
display(final_ranking)

best_predictive = final_ranking.iloc[0]
best_top_k = final_ranking.sort_values(
    ["Lift@10%", "Precision@10%", "Recall@10%"],
    ascending=False,
).iloc[0]

interpretable_candidates = final_ranking[
    final_ranking["Model"].isin(["LogisticRegression", "LinearSVC", "DecisionTreeClassifier"])
]
best_interpretable = interpretable_candidates.iloc[0]

print("\nRecommendations:")
print(f"Best interpretable model: {best_interpretable['Model']} | {best_interpretable['Feature set']}")
print(f"Best predictive model: {best_predictive['Model']} | {best_predictive['Feature set']}")
print(f"Best top-k screening model: {best_top_k['Model']} | {best_top_k['Feature set']}")
print(f"Suggested operating threshold: {best_predictive['Selected threshold']:.4f} (F1-optimized)")
print("Prototype limitations: synthetic/observational data, possible InterestRate underwriting leakage,")
print("SMOTENC probability distortion, no calibration, and no temporal/external validation.")
```
